```python
import os

from folioclient import FolioClient

okapi_url = os.getenv("OKAPI_URL")
user = os.getenv("OKAPI_USER")
password = os.getenv("OKAPI_PASSWORD")
tenant = os.getenv("TENANT")

folio_client = FolioClient(
    okapi_url,
    tenant,
    user,
    password)
```

```python
poline_id = "94067e0f-5a7c-43a3-9ca5-55d388ddb545"
order_line = folio_client.folio_get(f"/orders-storage/po-lines/{poline_id}")
is_package = order_line.get("isPackage")
print(is_package)
```

```python
if is_package is True:
    print(f"PO Line {poline_id} is for a package.")
else:
    print(f"PO Line {poline_id} is not for a package.")
```
